### Описание задачи:
Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

### Требования к базовому классу Task:
• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).
• Методы:
o MarkAsComplete(): метод для отметки задачи как выполненной.
o GetTaskDetails(): метод для получения деталей задачи.
o ReassignTo(): метод для переназначения задачи другому члену
команды.

### Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```

using System;
using System.Collections.Generic;

public interface IReportable
{
    string ReportFormat { get; set; }
    void GenerateReport();
}

public interface IAuditable
{
    DateTime LastAuditDate { get; set; }
    void RecordChange(string description);
}

public class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public int Priority { get; set; }
    public string Assignee { get; set; }
    public bool IsCompleted { get; set; }
    public DateTime CreatedAt { get; set; }
    public string Description { get; set; }
    public int EstimatedHours { get; set; }

    public Task(int id, string name, int priority, string assignee, int estimatedHours, string description)
    {
        TaskId = id;
        TaskName = name;
        Priority = priority;
        Assignee = assignee;
        EstimatedHours = estimatedHours;
        Description = description;
        CreatedAt = DateTime.Now;
        IsCompleted = false;
    }

    public virtual void MarkAsComplete()
    {
        if (IsCompleted)
        {
            Console.WriteLine($"Задача '{TaskName}' (ID: {TaskId}) уже была выполнена ранее!");
            return;
        }
        IsCompleted = true;
        Console.WriteLine($"Задача '{TaskName}' (ID: {TaskId}) выполнена.");
    }

    public virtual string GetTaskDetails()
    {
        string status = IsCompleted ? "Выполнена" : "В работе";
        return $"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | Исполнитель: {Assignee} | Статус: {status}";
    }

    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача '{TaskName}' переназначена: с '{Assignee}' на '{newAssignee}'");
        Assignee = newAssignee;
    }

    public void UpdatePriority(int newPriority)
    {
        Console.WriteLine($"Приоритет задачи '{TaskName}' изменен с {Priority} на {newPriority}");
        Priority = newPriority;
    }

    public void UpdateDescription(string newDesc)
    {
        Description = newDesc;
        Console.WriteLine($"Описание задачи '{TaskName}' обновлено.");
    }

    public DateTime GetEstimatedDate()
    {
        return CreatedAt.AddHours(EstimatedHours);
    }
    public virtual void MarkAsComplete(string comment)
    {
        MarkAsComplete();
        Console.WriteLine($"Комментарий: {comment}");
    }

    public virtual void ReassignTo(string newAssignee, string reason)
    {
        ReassignTo(newAssignee);
        Console.WriteLine($"Причина переназначения: {reason}");
    }

    public string GetTaskDetails(bool showDescription)
    {
        string details = GetTaskDetails();

        if (showDescription)
            details += $" | Описание: {Description}";

        return details;
}
    
}

public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string OriginalAssignee { get; set; }
    public int DelegationLevel { get; set; }

    public DelegateTask(int id, string name, int priority, string assignee, int estimatedHours, string description, DateTime dueDate)
        : base(id, name, priority, assignee, estimatedHours, description)
    {
        DueDate = dueDate;
        OriginalAssignee = assignee;
        DelegationLevel = 1;
    }

    public override void MarkAsComplete()
    {
        if (IsCompleted)
        {
            Console.WriteLine($"Делегированная задача '{TaskName}' (ID: {TaskId}) уже выполнена!");
            return;
        }
        base.MarkAsComplete();
        Console.WriteLine($"Делегированная задача выполнена. Уровень делегирования: {DelegationLevel}");
    }

    public void Escalate(string manager)
    {
        DelegationLevel++;
        Console.WriteLine($"Задача '{TaskName}' эскалирована менеджеру: {manager}");
        Assignee = manager;
    }

    public void CheckDeadline()
    {
        if (DateTime.Now > DueDate && !IsCompleted)
            Console.WriteLine($"ВНИМАНИЕ: Срок задачи '{TaskName}' истек!");
        else
            Console.WriteLine($"Задача '{TaskName}' в срок. Дедлайн: {DueDate.ToShortDateString()}");
    }

    public void ResetDelegation()
    {
        Assignee = OriginalAssignee;
        DelegationLevel = 1;
        Console.WriteLine($"Задача '{TaskName}' возвращена исходному исполнителю: {OriginalAssignee}");
    }
}

public class TeamTask : Task
{
    public string TeamName { get; set; }
    public string ProjectCode { get; set; }
    public string LeadDeveloper { get; set; }

    public TeamTask(int id, string name, int priority, string assignee, int estimatedHours, string description, string teamName, string projectCode)
        : base(id, name, priority, assignee, estimatedHours, description)
    {
        TeamName = teamName;
        ProjectCode = projectCode;
        LeadDeveloper = assignee;
    }

    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"В команде '{TeamName}' задача теперь числится за: '{newAssignee}'");
    }

    public void SetLead(string lead)
    {
        LeadDeveloper = lead;
        Console.WriteLine($"Назначен новый лид '{lead}' для команды '{TeamName}'");
    }

    public void ChangeProjectCode(string newCode)
    {
        ProjectCode = newCode;
        Console.WriteLine($"Код проекта изменен на: {ProjectCode}");
    }

    public void ShowTeamInfo()
    {
        Console.WriteLine($"Команда: {TeamName} | Код проекта: {ProjectCode} | Лид: {LeadDeveloper}");
    }
}

public class AgileTeamTask : TeamTask
{
    public int SprintNumber { get; set; }
    public int StoryPoints { get; set; }
    public string SprintStatus { get; set; }

    public AgileTeamTask(int id, string name, int priority, string assignee, int estimatedHours, string description, string teamName, string projectCode, int sprintNumber, int storyPoints)
        : base(id, name, priority, assignee, estimatedHours, description, teamName, projectCode)
    {
        SprintNumber = sprintNumber;
        StoryPoints = storyPoints;
        SprintStatus = "В работе";
    }

    public void StartSprint()
    {
        SprintStatus = "Активен";
        Console.WriteLine($"Спринт #{SprintNumber} для задачи '{TaskName}' запущен.");
    }

    public void CompleteSprint()
    {
        SprintStatus = "Завершен";
        base.MarkAsComplete();
        Console.WriteLine($"Спринт #{SprintNumber} завершен.");
    }

    public void ShowSprintInfo()
    {
        Console.WriteLine($"Спринт: {SprintNumber} | Story Points: {StoryPoints} | Статус: {SprintStatus}");
    }

    public override string GetTaskDetails()
    {
        return $"{base.GetTaskDetails()} | Спринт: {SprintNumber}";
    }
}

public class ResearchTask : Task
{
    public string DataSource { get; set; }
    public string ResearchMethod { get; set; }
    public bool IsPublished { get; set; }

    public ResearchTask(int id, string name, int priority, string assignee, int estimatedHours, string description, string dataSource, string researchMethod)
        : base(id, name, priority, assignee, estimatedHours, description)
    {
        DataSource = dataSource;
        ResearchMethod = researchMethod;
        IsPublished = false;
    }

    public override string GetTaskDetails()
    {
        return $"{base.GetTaskDetails()} | Источник: {DataSource}";
    }

    public void PublishFindings()
    {
        IsPublished = true;
        Console.WriteLine($"Результаты исследования '{TaskName}' опубликованы!");
    }

    public void ChangeSource(string newSource)
    {
        DataSource = newSource;
        Console.WriteLine($"Источник данных для '{TaskName}' обновлен на: {newSource}");
    }

    public void StartPeerReview()
    {
        Console.WriteLine($"Задача '{TaskName}' отправлена на рецензирование.");
    }
}

public class AdvancedResearchTask : ResearchTask, IReportable, IAuditable
{
    public string ReportFormat { get; set; }
    public DateTime LastAuditDate { get; set; }
    public int PageCount { get; set; }

    public AdvancedResearchTask(int id, string name, int priority, string assignee, int estimatedHours, string description, string dataSource, string researchMethod, string reportFormat)
        : base(id, name, priority, assignee, estimatedHours, description, dataSource, researchMethod)
    {
        ReportFormat = reportFormat;
        LastAuditDate = DateTime.Now;
        PageCount = 0;
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Генерация отчета по задаче '{TaskName}' в формате {ReportFormat}. Страниц: {PageCount}");
        RecordChange("Сгенерирован отчет");
    }

    public void RecordChange(string description)
    {
        LastAuditDate = DateTime.Now;
        Console.WriteLine($"[АУДИТ] {LastAuditDate}: {description} для задачи '{TaskName}'");
    }

    public void UpdatePageCount(int pages)
    {
        PageCount = pages;
        Console.WriteLine($"Количество страниц в исследовании '{TaskName}' обновлено: {PageCount}");
        RecordChange($"Обновление страниц до {PageCount}");
    }
}


public class TaskCollection<T> where T : Task
{
    private List<T> tasks = new List<T>();

    public void Add(T task)
    {
        tasks.Add(task);
    }

    public void Remove(T task)
    {
        tasks.Remove(task);
    }

    public void ShowTasks()
    {
        foreach (T task in tasks)
        {
            Console.WriteLine(task.GetTaskDetails());
        }
    }
}

TaskCollection<Task> collection = new TaskCollection<Task>();

Task task1 = new DelegateTask(
    1,
    "Подготовить отчет",
    1,
    "Иван",
    5,
    "Создать отчет",
    DateTime.Now.AddDays(3));

Task task2 = new TeamTask(
    2,
    "Разработать модуль",
    2,
    "Анна",
    8,
    "Создать программный модуль",
    "Backend",
    "PRJ-001");

collection.Add(task1);
collection.Add(task2);

// Полиморфизм с перекрытием
Task[] tasks = { task1, task2 };

foreach (Task task in tasks)
{
    Console.WriteLine(task.GetTaskDetails());
}

// Перегрузка методов
task1.MarkAsComplete("Проверено руководителем");
task2.ReassignTo("Петр", "Изменился исполнитель");

Console.WriteLine(task1.GetTaskDetails(true));

// Generic-класс
Console.WriteLine("\nВсе задачи:");
collection.ShowTasks();
```
