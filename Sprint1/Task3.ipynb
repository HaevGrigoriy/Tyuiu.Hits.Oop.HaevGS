### Описание задачи:
Создать базовый класс Task в C#, который будет представлять задачи внутри
проекта. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

### Требования к базовому классу Task:
• Атрибуты: ID задачи (TaskId), Название задачи (TaskName), Приоритет
задачи (Priority).
• Методы:
o MarkAsComplete(): метод для отметки задачи как выполненной.
o GetTaskDetails(): метод для получения деталей задачи.
o ReassignTo(): метод для переназначения задачи другому члену
команды.

### Требования к производным классам:
1. ДелегатскаяЗадача (DelegateTask): Должна содержать дополнительные
атрибуты, такие как Дата выполнения (DueDate).
Метод MarkAsComplete() должен быть переопределен для включения даты
выполнения в сообщение о завершении задачи.
2. КоманднаяЗадача (TeamTask): Должна содержать дополнительные атрибуты,
такие как Команда (TeamName). Метод ReassignTo() должен быть
переопределен для указания нового члена команды, которому будет
переназначена задача.
3. ИсследовательскаяЗадача (ResearchTask) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Исходные данные
(DataSource). Метод GetTaskDetails() должен быть переопределен для
отображения источников данных, используемых в задаче, вместе с другими
деталями задачи.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

```
class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public int Priority { get; set; }
    public string Assignee { get; set; }
    public bool IsCompleted { get; set; }

    public Task() { }


    public Task(int id, string name, int priority, string assignee)
    {
        TaskId = id;
        TaskName = name;
        Priority = priority;
        Assignee = assignee;
    }

    public virtual void MarkAsComplete()
    {

        if (IsCompleted)
        {
            Console.WriteLine($"Задача '{TaskName}' (ID: {TaskId}) уже была выполнена ранее!");
            return;
        }

        IsCompleted = true;
        Console.WriteLine($"Задача {TaskName}, (ID:{TaskId}) выполнена");
    }

    public virtual string GetTaskDetails()
    {
        string status = IsCompleted ? "Выполнена" : "В работе";
        return $"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | Исполнитель: {Assignee} | Статус: {status}";
    }

    public virtual void ReassignTo(string newAssignee)
    {
        Console.WriteLine($"Задача '{TaskName}' переназначена: с '{Assignee}' на '{newAssignee}'");
        Assignee = newAssignee;
    }
}

class DelegateTask : Task
{
    public DateTime DueDate { get; set; }

    public DelegateTask() { }

    public DelegateTask(int id, string name, int priority, string assignee, DateTime dueDate)
        : base(id, name, priority, assignee)
    {
        DueDate = dueDate;
    }

    public override void MarkAsComplete()
    {
        if (IsCompleted)
        {
            Console.WriteLine($"Делегированная задача '{TaskName}' (ID: {TaskId}) уже выполнена!");
            return;
        }

        base.MarkAsComplete();
        Console.WriteLine($"Делегировання задача '{TaskName}', (ID: {TaskId}) выполнена в срок. Плановая дата выполнения (DueDate): {DueDate.ToShortDateString()}");
    }
}

class TeamTask : Task
{
    public string TeamName { get; set; }

    public TeamTask() { }

    public TeamTask(int id, string name, int priority, string assignee, string teamName)
        : base(id, name, priority, assignee)
    {
        TeamName = teamName;
    }

    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"В команде '{TeamName}' задача теперь будет выполнять сотрудник: '{newAssignee}'");
    }
}

class ResearchTask : Task
{
    public string DataSource { get; set; }

    public ResearchTask() { }

    public ResearchTask(int id, string name, int priority, string assignee, string dataSource)
        : base(id, name, priority, assignee)
    {
        DataSource = dataSource;
    }

    public override string GetTaskDetails()
    {
        string baseDetails = base.GetTaskDetails();
        return $"{baseDetails} | Источник данных: {DataSource}";
    }
}

Task myTask = new Task();
DelegateTask myDelegateTask = new DelegateTask();
TeamTask myTeamTask = new TeamTask();
ResearchTask myResearchTask = new ResearchTask();

myTask.TaskId = 1;
myTask.TaskName = "Спринт 0";
myTask.Priority = 5;
myTask.Assignee = "ТИУ";
myTask.IsCompleted = false; 
myTask.MarkAsComplete();
Console.WriteLine(myTask.GetTaskDetails());
myTask.ReassignTo("Хаев Григорий");

myTask.MarkAsComplete(); 

myDelegateTask.TaskId = 2;
myDelegateTask.TaskName = "Ревью кода";
myDelegateTask.Priority = 3;
myDelegateTask.Assignee = "Петров";
myDelegateTask.DueDate = DateTime.Now.AddDays(5);
myDelegateTask.MarkAsComplete();

myTeamTask.TaskId = 3;
myTeamTask.TaskName = "Тестирование";
myTeamTask.Priority = 4;
myTeamTask.Assignee = "Сидоров";
myTeamTask.TeamName = "ТОПППППП";
        
myTeamTask.ReassignTo("Хаев Григорий");

myResearchTask.TaskId = 4;
myResearchTask.TaskName = "Исследование API";
myResearchTask.Priority = 2;
myResearchTask.Assignee = myTeamTask.Assignee;
myResearchTask.DataSource = "GitHub";

Console.WriteLine(myTask.GetTaskDetails());
Console.WriteLine(myResearchTask.GetTaskDetails());
```
